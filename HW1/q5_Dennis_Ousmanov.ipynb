```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
n = 1000000
sh_analysis = stats.expon.rvs(scale=10, size=n)
extr = stats.norm.rvs(loc=15, scale=3, size=n)
eda = stats.norm.rvs(loc=3, scale=0.5, size=n)
modl = stats.poisson.rvs(mu=30, size=n)
process_int = stats.norm.rvs(loc=50, scale=20, size=n)
pipe = stats.poisson.rvs(mu=100, size=n)
modl_prod = stats.norm.rvs(loc=30, scale=5, size=n)
```

```python
# Without Stakeholder analysis we cannot do development + pipelline
# No steps with EDA or after without extraction
devel = eda + modl + modl_prod
prod = pipe
print((devel > prod).sum()/n)
# percent where development takes longer than production
```

1.

```python
days = sh_analysis + extr + np.amax(np.column_stack((devel,prod)),1) + process_int

print("Mean number of days for process:", days.mean())
print("Median number of days for process:", np.median(days))
```

```python
print("two sided confidence interval:", np.quantile(days, [0.025, 0.975]))
```

2.

```python
print("Variance of production days:", np.var(prod))
print("Mean of production days:", np.mean(prod))
print("Variance of development days:", np.var(devel))
print("Mean of development days:", np.mean(devel))
print("Variance of process integration days:", np.var(process_int))
print("Mean of process integration days:", process_int.mean())
print("Variance of stakeholder analysis days:", np.var(sh_analysis))
print("Mean of stakeholder analysis days:", sh_analysis.mean())
print("Variance of extraction days:", np.var(extr))
print("Mean of extraction days:", extr.mean())
```

The variance is the highest for the Process Integration step, which can only be done after the Production Pipeline and model productionization are complete. Since it has the highest variance in days units, it contributes the most towards process time uncertaininty. 

The data pipeline contributes most to the overall duration because its mean is 100 days and is almost always the bottleneck in the dev vs prod step.

```python
# So days represents our 1 million simulations of the total days for the process withou delay
additional_pipe_time = stats.poisson.rvs(50,size=n)
prod_with_delay = pipe + additional_pipe_time 

days_w_delay = sh_analysis + extr + np.amax(np.column_stack((devel,prod_with_delay)),1) + process_int

length_of_delay = (days_w_delay - days)
print("Mean length of delay:", length_of_delay.mean())
print("two sided confidence interval for length of delay:", np.quantile(length_of_delay, [0.025, 0.975]))
```
