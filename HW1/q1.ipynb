```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('sampregdata.csv', index_col = 0)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X = df.drop(columns = ['y'])
y = df['y']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

```python
features = X_train.columns
min_mse = float('inf')

for feature in features:
    candidate_model = LinearRegression()
    candidate_model.fit(X_train[[feature]], y_train)
    y_pred = candidate_model.predict(X_test[[feature]])
    curr_mse = mean_squared_error(y_test, y_pred)

    if curr_mse < min_mse:
        min_mse = curr_mse
        best_feature = feature
        best_one_feature_model = candidate_model
    
print(f"Best feature: {best_feature}, Best MSE: {min_mse}")
```

```python
features_reduced = X_train.drop(columns = [best_feature]).columns
min_mse = float('inf')

for feature in features_reduced:
    features_to_use = [best_feature, feature]
    candidate_model = LinearRegression()
    candidate_model.fit(X_train[features_to_use], y_train)
    y_pred = candidate_model.predict(X_test[features_to_use])
    curr_mse = mean_squared_error(y_test, y_pred)

    if curr_mse < min_mse:
        min_mse = curr_mse
        best_features = features_to_use
        best_two_feature_model = candidate_model
    
print(f"Best feature: {best_features}, Best MSE: {min_mse}")
```

```python
# sanity check that the models are accessible 
pred_one_feature = best_one_feature_model.predict(X_test[[best_feature]])
pred_two_features = best_two_feature_model.predict(X_test[best_features])

print(np.isnan(pred_one_feature).sum())
print(np.isnan(pred_two_features).sum())
```
