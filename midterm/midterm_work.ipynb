```python
import numpy as np

rng = np.random.default_rng(890)
n = 1_000_000

X = rng.uniform(5, 10, n)
Y = rng.uniform(1, 20, n)

error_X = rng.normal(0, 2, n)
error_Y = rng.normal(0, 0.1, n)

X_pred = X + error_X
Y_pred = Y + error_Y
```

```python
Z_actual = X / Y
Z_pred = X_pred / Y_pred
Z_error = Z_pred - Z_actual

print(f"Mean error: {Z_error.mean():.6f}")
print(f"Standard deviation of error: {Z_error.std(ddof=1):.6f}")
```
